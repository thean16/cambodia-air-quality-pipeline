```python
import requests
import pandas as pd
import os
import time
from datetime import datetime

# NEW — saves inside your current VS Code project folder
import os
OUTPUT_DIR = os.path.join(os.getcwd(), "live_input")
print("Saving CSVs to:", OUTPUT_DIR)
INTERVAL   = 60  # seconds between each CSV

CITIES = [
    {"city": "Phnom Penh",    "lat": 11.5564, "lon": 104.9282},
    {"city": "Siem Reap",     "lat": 13.3671, "lon": 103.8448},
    {"city": "Battambang",    "lat": 13.1024, "lon": 103.1988},
    {"city": "Kampot",        "lat": 10.6108, "lon": 104.1812},
    {"city": "Sihanoukville", "lat": 10.6097, "lon": 103.5297},
]
```

```python
def fetch_one_city(city_info):
    url = "https://air-quality-api.open-meteo.com/v1/air-quality"
    params = {
        "latitude":  city_info["lat"],
        "longitude": city_info["lon"],
        "current":   "pm2_5,pm10,carbon_monoxide,nitrogen_dioxide,sulphur_dioxide,ozone,european_aqi",
        "timezone":  "Asia/Phnom_Penh",
    }
    try:
        res = requests.get(url, params=params, timeout=15)
        res.raise_for_status()           # crash if HTTP error
        current = res.json()["current"]

        return {
            "city":       city_info["city"],
            "latitude":   city_info["lat"],
            "longitude":  city_info["lon"],
            "datetime":   current.get("time"),
            "aqi":        current.get("european_aqi"),
            "pm25":       current.get("pm2_5"),
            "pm10":       current.get("pm10"),
            "co":         current.get("carbon_monoxide"),
            "no2":        current.get("nitrogen_dioxide"),
            "so2":        current.get("sulphur_dioxide"),
            "o3":         current.get("ozone"),
            "created_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        }

    except requests.exceptions.Timeout:
        print(f"  [TIMEOUT]  {city_info['city']}")
    except requests.exceptions.ConnectionError:
        print(f"  [NO INTERNET]  {city_info['city']}")
    except Exception as e:
        print(f"  [ERROR]  {city_info['city']} — {e}")

    return None   # skip this city if any error
```

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ONE fixed filename — never changes
COMBINED_FILE = os.path.join(OUTPUT_DIR, "air_quality_combined.csv")

while True:
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    print(f"\n[{timestamp}] Fetching data...")

    rows = []
    for city in CITIES:
        result = fetch_one_city(city)
        if result:
            rows.append(result)
            print(f"  ✓ {city['city']}  AQI={result['aqi']}  PM2.5={result['pm25']}")

    if rows:
        new_df = pd.DataFrame(rows)

        # If file exists → append, if not → create with header
        if os.path.exists(COMBINED_FILE):
            new_df.to_csv(COMBINED_FILE, mode="a", index=False, header=False)
            print(f"  → Appended {len(rows)} rows to air_quality_combined.csv")
        else:
            new_df.to_csv(COMBINED_FILE, mode="w", index=False, header=True)
            print(f"  → Created air_quality_combined.csv")

        # Show current total rows
        total = pd.read_csv(COMBINED_FILE).shape[0]
        print(f"  → Total rows in file: {total}")

    print(f"  Waiting {INTERVAL}s...")
    time.sleep(INTERVAL)
```
